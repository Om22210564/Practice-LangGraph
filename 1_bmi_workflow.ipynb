```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BMIstate(TypedDict):
    weight: float
    height: float
    bmi: float
```

```python
def calculateBMI(state:BMIstate) -> BMIstate:
    state['bmi'] = state['weight']/state['height']**2
    return state
```

```python
# to define the graph
graph = StateGraph(BMIstate)

# add node
graph.add_node('calculate_BMI',calculateBMI)

# add edge
graph.add_edge(START,'calculate_BMI')
graph.add_edge('calculate_BMI',END)

# compile the graph
worfklow = graph.compile()
```

```python
worfklow
```

```python
Initial_state= {"weight": 100, "height": 1.2}
```

```python
Final_state= worfklow.invoke(Initial_state)
```
